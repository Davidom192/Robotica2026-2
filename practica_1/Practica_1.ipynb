###2. ¿Qué es un archivo URDF?

Un archivo URDF (Unified Robot Description Format) es un formato basado en XML que se usa para describir la estructura física de un robot. Es muy común en robótica, especialmente con ROS.

Con un archivo URDF puedes decirle al sistema cosas como:


*   Partes del robot (Eslabones)
*   Cómo se conectan (Juntas)
*   Dimensiones
*   Masas e inercias
*   Elementos esteticos como colores

###4. ¿Que es RVIZ?

RViz (abreviatura de ROS Visualization) es una herramienta de visualización 3D diseñada para el entorno del sistema operativo robótico ROS. Permite a los desarrolladores e investigadores inspeccionar en tiempo real los datos de sensores, mapas, modelos de robots y trayectorias en entornos simulados o reales.

###2. ¿Cuál es la convención de ángulos que utilizan los archivos URDF para los ángulos de las juntas respecto al eslabón padre?


En los archivos URDF, los ángulos de las juntas no usan directamente una convención tipo Euler fija para definir el movimiento de la junta, sino que la orientación respecto al eslabón padre se define mediante el elemento origin.

Aquí la convención es RPY (Roll–Pitch–Yaw) y sigue este orden:

Roll (r) → rotación sobre eje X
Pitch (p) → rotación sobre eje Y
Yaw (y) → rotación sobre eje Z

Los ángulos están en radianes.

##**Conclusiones**


A partir del desarrollo de esta práctica, se logró comprender, interpretar y modificar de manera exitosa la estructura fundamental de un archivo URDF (Universal Robot Description Format). La manipulación de los elementos <link> y <joint> permitió asimilar cómo se configuran las propiedades geométricas, visuales y cinemáticas de un robot serial (en este caso, un manipulador tipo SCARA), así como la importancia de dominar la convención RPY (Roll-Pitch-Yaw) en radianes para establecer las transformaciones espaciales correctas respecto a los eslabones padre.

La visualización en tiempo real dentro del entorno de RViz fue una herramienta clave para validar de forma gráfica el diseño del robot. Esta etapa permitió corroborar que las dimensiones de los eslabones, los límites de las articulaciones (tanto revolutas como prismáticas) y la alineación de los sistemas de referencia declarados en el código XML correspondieran fielmente al comportamiento físico esperado del sistema mecatrónico.

Finalmente, la integración de herramientas de control de versiones como Git y GitHub reforzó la competencia en metodologías de trabajo actuales y profesionales.
